```python
# Multiple conditions - AND
df[(df['Category'] == 'Furniture') & (df['Price'] > 100)]
```

```python
# Multiple conditions - OR
df[(df['Category'] == 'Clothing') | (df['Price'] > 100)]
```

```python
# Multiple conditions - Comparison operators
df[df['Quantity'] > 10] # uther comparison operators: <, <=, >, >=, !=
```

```python
# String methods
df[df['CustomerName'].str.startswith('A')] # other string methods: endswith(), contains(), len(), lower(), upper()
```

```python
# boolean/conditional/predicate methods.
df[df['Category'].isin(['Furniture', 'Clothing'])] # other methods: notin(), between(), isnull(), notnull()
```

```python
# ~ Negation operator
df[~df['Country'].isin(['United States', 'Canada', 'Mexico'])] # other boolean operators: &, |, ^, ==, !=, >, <, >=, <= 
```

```python
# Updating values: increase price by 10% for furniture items
df.loc[df['Category'] == 'Furniture', 'Price'] = df['Price'] * 1.1 
```

```python
# update product for Anna Ivanova to Office Chair
df.loc[df['CustomerName'] == 'Anna Ivanova', 'Product'] = 'Office Chair'
```

```python
# Display the updated row for Anna Ivanova
df.loc[df['CustomerName'] == 'Anna Ivanova'] 
```

```python
# Convert country names to uppercase for all rows
df['Country'] = df['Country'].str.upper()
```

```python
# Delete a row by index
df = df.drop(39) 
```

## aggregation/statistical methods in pandas. They work on a numeric Series, not on a string column
* `mean()`   → the average value.
* `median()`  → the middle value after sorting. For number values only.
* `mode()`  → the most frequent value. Works for strings too.
* `std()`  → standard deviation.  For number values only.
* `var()`  → variance.  For number values only.
* `min()`  → the minimum value.
* `max()`  → the maximum value.
